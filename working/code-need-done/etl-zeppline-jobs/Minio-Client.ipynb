```python
from minio import Minio
import urllib3
import os
# Tạo một HttpClient tùy chỉnh bỏ qua xác thực SSL
http_client = urllib3.PoolManager(
    cert_reqs='CERT_NONE',
    assert_hostname=False
)

client = Minio(
    os.getenv("MINIO_ENDPOINT"),        # Gọi qua IP Nginx cổng 9000
    access_key=os.getenv("MINIO_USERNAME"),
    secret_key=os.getenv("MINIO_PASSWORD"),
    secure=True,                  # Để False vì bạn đang dùng HTTP serving,
    http_client=http_client,
)
```

```python
# Thử tạo một Bucket mới
if not client.bucket_exists("backend-data"):
    client.make_bucket("backend-data")
    print("Tạo bucket thành công!")
```

```python
# Thử tạo một Bucket mới
if client.bucket_exists("backend-data"):
    v = client.remove_bucket("backend-data")
    
    print("Xoa bucket thành công!", v)
```

```python

bucket_name = "jira"
# Đường dẫn bạn muốn lưu trên MinIO (Object Name)
object_name = "projects/v1/documents/test-file.txt" 
# Đường dẫn file trên máy tính của bạn
file_path = r"C:\Users\namtv40\Documents\Crawlers\Noc\Queries.xlsx"

# 1. Đảm bảo Bucket tồn tại
if not client.bucket_exists(bucket_name):
    client.make_bucket(bucket_name)

# 2. Upload file
# MinIO tự động "tạo" các folder: projects, v1, documents
client.fput_object(
    bucket_name, 
    object_name, 
    file_path
)

print(f"Thành công! File đã nằm tại: {bucket_name}/{object_name}")
# Tạo một folder trống tên là 'templates'
# client.put_object(bucket_name, "templates/", data=b"", length=0)
```
